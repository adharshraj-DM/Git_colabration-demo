## 1. What is Pandas?

**Definition:** Pandas is a Python library used to work with **tabular data** — data arranged in rows and columns.

Think of Pandas as a **smart Excel sheet inside Python**.

### Two important Pandas objects

| Object | Simple meaning |
|---|---|
| **Series** | One column of data |
| **DataFrame** | A complete table with rows and columns |

Pandas is built on top of NumPy and is widely used for data analysis and data cleaning.

````python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
````

# PART 1 — SERIES

## 2. What is a Series?

A **Series** is a one-dimensional labeled data structure.

### Easy way to remember

```text
Series = ONE column
```

Example:

```text
0    10
1    20
2    30
3    40
```

The numbers on the left are the **index** and the numbers on the right are the **values**.

## 3. Creating a Series from a tuple

````python
t = (10, 11, 12)

s = pd.Series(t)

s
````

## 4. Creating a Series from a list

````python
marks = [45, 78, 56, 89, 72]

s = pd.Series(marks)

s
````

## 5. Creating a Series from a NumPy array

````python
arr = np.array([1, 2, 3, 4])

s = pd.Series(arr)

s
````

## 6. Creating a Series from a dictionary

A dictionary is especially useful because:

```text
dictionary key   → Series index
dictionary value → Series value
```

````python
d = {
    'a': 1,
    'b': 2,
    'c': 3,
    'd': 4
}

s = pd.Series(d)

s
````

## 7. Series indexing

We can access a value using its index.

```python
s['a']
```

We can also access a value by position.

```python
s.iloc[2]
```

````python
print("Value at label 'a':", s['a'])
print("Value at position 2:", s.iloc[2])
````

## 8. Series slicing

### Important

For normal Python/Pandas positional slicing:

```python
series[start:stop]
```

the **stop position is excluded**.

Example:

```python
s[1:5]
```

means:

```text
start at 1
stop before 5
```

````python
s = pd.Series(np.arange(10))

print("Original Series:")
print(s)

print("\nSlice 1:5:")
print(s[1:5])
````

# PART 2 — DATAFRAME

## 9. What is a DataFrame?

A **DataFrame** is a two-dimensional labeled data structure.

### Easy way to remember

```text
Series    = ONE column
DataFrame = MANY columns + MANY rows
```

Think of a DataFrame as a table:

| Name | Age |
|---|---:|
| A | 25 |
| B | 30 |
| C | 22 |

## 10. Creating a DataFrame from nested lists

````python
data = [
    ['abc', 1],
    ['def', 2],
    ['ghi', 3],
    ['hij', 4],
    ['klm', 5]
]

df = pd.DataFrame(data)

df
````

## 11. Giving column names

Instead of default column numbers, we can provide meaningful names.

````python
df = pd.DataFrame(
    data,
    columns=['Name', 'Number']
)

df
````

## 12. Giving custom index labels

````python
df = pd.DataFrame(
    data,
    columns=['Name', 'Number'],
    index=['A', 'B', 'C', 'D', 'E']
)

df
````

## 13. Creating a DataFrame from a dictionary

This is one of the most common ways to create a DataFrame.

```text
Dictionary key   → column name
Dictionary list  → column values
```

````python
data = {
    'name': ['a', 'b', 'c', 'd', 'e', 'f'],
    'age': [45, 12, 48, 25, 12, 56]
}

df = pd.DataFrame(data)

df
````

## 14. Renaming columns

### Method 1 — Rename all columns

````python
df.columns = ['Candidate', 'Age_in_years']

df
````

### Method 2 — Rename only a specific column

`axis=1` means **columns**.

````python
df.rename(
    {'Age_in_years': 'Age'},
    axis=1,
    inplace=True
)

df
````

### What does `inplace=True` mean?

- `inplace=False` → create/return a changed version
- `inplace=True` → change the existing DataFrame itself

Easy memory trick:

```text
inplace=True → "Change this DataFrame directly."
```

# PART 7 — `loc`

## 30. What is `loc`?

`loc` is used for **label-based selection** and, very commonly, **conditional filtering**.

Think:

```text
loc = "Look for rows that satisfy my condition."
```

````python
# Find the passenger whose age is exactly 80
data.loc[data['Age'] == 80]
````

## 31. `loc` with a condition

## 32. `loc` with a categorical condition

# PART 8 — `iloc`

## 33. What is `iloc`?

`iloc` means **integer-location based selection**.

Think:

```text
loc  → labels / conditions
iloc → positions (integer positions)
```

Example:

```python
data.iloc[0]
```

means:

> Give me the row at position 0.

````python
data.iloc[0]
````

## 34. `iloc` for rows and columns

Syntax:

```python
data.iloc[row_start:row_stop, column_start:column_stop]
```

The stop position is excluded in positional slicing.

Example:

```python
data.iloc[15:30, 1:8]
```

````python
data.iloc[15:30, 1:8]
````

## 35. Using `np.where()` with `iloc`

The original teaching notebook also demonstrates this pattern:

```python
data.iloc[np.where(condition)]
```

It first finds the positions satisfying the condition and then uses `iloc` to select those positions.

````python
data.iloc[np.where(data['Age'] == 80)]
````

# PART 9 — MULTIPLE CONDITIONS

## 36. AND condition: `&`

Use `&` when **both conditions must be true**.

Question:

> Find females whose age is less than 38.

````python
data.loc[
    (data['Sex'] == 'female') &
    (data['Age'] < 38)
]
````

### Mental model

```text
Female AND Age < 38
       ↓
Both conditions must be TRUE
```

⚠️ Put each condition inside parentheses.

## 37. OR condition: `|`

Use `|` when **at least one condition must be true**.

Question:

> Find passengers who are female OR younger than 38.

````python
data.loc[
    (data['Sex'] == 'female') |
    (data['Age'] < 38)
]
````

### Mental model

```text
Female OR Age < 38
      ↓
At least one condition must be TRUE
```

## 38. Combining three conditions

Question:

> Find female passengers in first class who survived.

````python
data.loc[
    (data['Sex'] == 'female') &
    (data['Pclass'] == 1) &
    (data['Survived'] == 1)
]
````

## 39. Another useful filter

Question:

> Find males who survived.

````python
data.loc[
    (data['Sex'] == 'male') &
    (data['Survived'] == 1)
]
````

# PART 10 — MISSING VALUES

## 40. What is a missing value?

A missing value means the dataset does not contain a value for that particular cell.

Pandas commonly represents missing numerical/text values as `NaN`.

The Titanic dataset contains missing values, especially in `Age`, `Cabin`, and `Embarked`.

````python
print("Missing values in Age:")
print(data['Age'].isnull().sum())

print("\nMissing values in every column:")
print(data.isnull().sum())
````

## 41. `isnull()`

```python
data.isnull()
```

returns `True` where a value is missing and `False` where it is not.

Adding `.sum()` counts the missing values.

```python
data.isnull().sum()
```

### Easy memory trick

```text
isnull()       → find missing cells
isnull().sum() → count missing cells
```

# PART 15 — SORTING

## 51. `sort_values()`

Sorting means arranging rows according to a column.

Question:

> Arrange passengers by passenger class.

````python
data.sort_values('Pclass').head(10)
````

## 52. Descending order

```python
ascending=False
```

means largest/highest values first.

````python
data.sort_values('Pclass', ascending=False).head(10)
````

## 53. `ignore_index=True`

After sorting, the original row index may remain.

`ignore_index=True` creates a fresh `0, 1, 2, ...` index for the result.

````python
data.sort_values(
    'Pclass',
    ascending=False,
    ignore_index=True
).head(10)
````

## 54. Sorting by multiple columns

We can sort by more than one column.

Example:

```python
by=['Age', 'Pclass']
ascending=[True, False]
```

means:

1. Sort Age from low to high.
2. If ages are equal, sort Pclass from high to low.

````python
data.sort_values(
    by=['Age', 'Pclass'],
    ascending=[True, False]
).head(20)
````

# PART 16 — GROUPBY

## 55. What is `groupby()`?

`groupby()` means:

> **Divide the data into groups and then perform an operation on each group.**

Imagine separating Titanic passengers into:

```text
Pclass = 1
Pclass = 2
Pclass = 3
```

Then we can calculate something for each group.

````python
data.groupby('Pclass').size()
````

## 56. Group by Pclass and calculate average Age

This is a very useful and beginner-friendly groupby example.

````python
data.groupby('Pclass')['Age'].mean()
````

## 57. Group by Pclass and find maximum Fare

````python
data.groupby('Pclass')['Fare'].max()
````

## 58. `agg()`

`agg()` lets us specify the aggregation we want.

Example:

```python
groupby('Pclass').agg({'Fare': 'max'})
```

````python
data.groupby('Pclass').agg({'Fare': 'max'})
````

## 59. Group by Sex

We can also group by a categorical column.

For example, calculate average age for each sex:

````python
data.groupby('Sex')['Age'].mean()
````

### Important note about the original notebook

The original material contains:

```python
data.groupby(data['Sex']).mean()
```

For modern Pandas versions, it is clearer and safer to specify the numeric column explicitly:

```python
data.groupby('Sex')['Age'].mean()
```

The teaching idea is the same: **group first, then calculate an aggregate.**

# PART 17 — CHANGING DATA TYPES

## 60. Why change data types?

A DataFrame contains different kinds of data:

```text
PassengerId → integer
Age         → floating point
Name        → text/object
Sex         → text/object
```

Sometimes we need to convert a column to another type.

````python
data.info()
````

## 61. `astype()`

`astype()` converts a column to a specified data type.

Example from the original teaching material:

```python
data['Age'] = data['Age'].astype('float16')
```

⚠️ We will demonstrate this on a copy so the original dataset stays unchanged.

````python
dtype_demo = data.copy()

dtype_demo['Age'] = dtype_demo['Age'].astype('float16')

dtype_demo[['Age']].head()
````

## 62. `pd.to_numeric()`

`pd.to_numeric()` converts values to numeric form.

`downcast` can request a smaller numeric representation when possible.

````python
dtype_demo = data.copy()

dtype_demo['Fare'] = pd.to_numeric(
    dtype_demo['Fare'],
    downcast='float'
)

dtype_demo['Parch'] = pd.to_numeric(
    dtype_demo['Parch'],
    downcast='integer'
)

dtype_demo.info()
````
